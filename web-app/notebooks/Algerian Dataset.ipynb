### About the dataset

1. Temperature in Celcuis: 22 to 42
2. RH - Relative Humidity in %: 21 to 90
3. Ws - Wind speed in Km/hr: 6-29
4. FFMC- Fine Fuel Moisture Code (FFMC) index from the FWI system: 28.6 to 92.5
5. DMC - Duff Moisture Code: 1.1 to 65.9
6. DC - Drought Code index: 7-220.4
7. ISI - Initial Spread Index:0-18.5
8. BUI - BUildup Index from the FWI system: 1.1 - 68
9. FWI - Fire Weather Index: 0-31.1
10. Classes - Two classes named Fire and Not Fire

# Data Cleaning

```python
## Missing values
dataset.isnull()
```

```python
dataset[dataset.isnull().any(axis = 1)]
```

```python
dataset.drop(index = 122)
```

The data is seperated between two parts
1. Bejaia Region Dataset
2. Sidi-Bel Abbes Region Dataset

```python
dataset.loc[:122, "Region"] = 0
dataset.loc[122:, "Region"] = 1
```

```python
dataset.head()
```

```python
dataset.info()
```

```python
dataset[['Region']] = dataset[["Region"]].astype(int)
```

```python
dataset.head()
```

```python
## Remove the null values
dataset = dataset.dropna().reset_index(drop = True)
dataset.head()
```

```python
dataset.iloc[[122]]
```

```python
dataset = dataset.drop(122).reset_index(drop = True)
dataset.iloc[[122]]
```

```python
dataset.columns
```

```python
## fix the space in columns name
dataset.columns = dataset.columns.str.strip()
dataset.columns
```

```python
dataset.info()
```

## change the required columns as integer datatypw

```python
dataset[["month","day","year","Temperature", "RH","Ws"]] = dataset[["month","day","year","Temperature", "RH","Ws"]].astype(int)
```

## Changing the other column as float datatype

```python
objects = [features for features in dataset.columns if dataset[features].dtype == "O"]
```

```python
for i in objects:
    if i != "Classes":
        dataset[i] = dataset[i].astype(float)
```

```python
dataset.info()
```

### Since Classes is Binary

### Since Classes in object: Binary: Fire and not-fire

```python
##Let save the cleaned file
dataset.to_csv("Algerian_forest_fires_cleaned_dataset.csv", index = False)
```

```python
## Monthly data analysis
# for bejaiya region

df['Classes'] = np.where(df['Classes'].str.contains('not fire'), 'not fire', 'fire')
dftemp = df.loc[df['Region']== 0]
plt.subplots(figsize=(13,6))
sns.set_style("whitegrid")
sns.countplot(x='month', hue = 'Classes', data=df)
plt.xlabel('Months' ,weight = 'bold')
plt.ylabel('Numbeer of event' ,weight = 'bold')
plt.title("Monthly fire distribution of Bijaya region")
plt.show()
```

```python
## Monthly data analysis
# for Sidi-Bel Abbes region


dftemp = df.loc[df['Region']== 1]
plt.subplots(figsize=(13,6))
sns.set_style("whitegrid")
sns.countplot(x='month', hue = 'Classes', data=df)
plt.xlabel('Months' ,weight = 'bold')
plt.ylabel('Numbeer of event' ,weight = 'bold')
plt.title("Monthly fire distribution of Sidi-Bel Abbes region")
plt.show()
```
